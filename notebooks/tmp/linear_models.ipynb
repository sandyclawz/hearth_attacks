#### Унифицируем все категориальные значения во всех датафреймах

````python
# Красиво выводит уникальные значения указанного столбца датафрейма.
def show_unique_values(df, column_name, df_name='DataFrame', emoji='🐮'):
    """
    Параметры:
    df (pd.DataFrame): датафрейм
    column_name (str): название столбца
    df_name (str): имя датафрейма (для отображения в заголовке)
    emoji (str): иконка для оформления (по умолчанию 🐮)
    """
    print(f"{emoji} Уникальные значения в поле '{column_name}' "
          f"в датафрейме {df_name}:\n")
    
    unique_vals = df[column_name].unique()
    
    for i, val in enumerate(unique_vals, 1):
        print(f"{i:>2}. {val}")
    
    print(f"\nВсего уникальных значений: {len(unique_vals)} \n-------------\n")
````

#### Проведем объединение датафреймов и создадим общие ключи

**🧬 Создание уникального идентификатора отца (`id_father`)**

В некоторых случаях одно и то же имя отца может встречаться у коров **разных пород**,  
поэтому необходимо сформировать уникальный код отца на основе **двух полей**:
- `father_name` — имя отца,  
- `father_breed` — порода отца.

````python
# создаём уникальный идентификатор по комбинации двух полей
cow_buy['id_father'] = cow_buy.groupby(['father_breed', 'father_name']).ngroup() + 1

# проверим результат
display(cow_buy[['father_breed', 'father_name', 'id_father']].head(20).sort_values('id_father'))
````

````python
### 🔗 Объединение датафреймов `ferma_main` и `ferma_dad`
print ("Длина датафрейма ferma_main до добавления информации из датафрейма ferma_dad: ", len(ferma_main))
ferma_main = ferma_main.merge(
    ferma_dad[['id', 'father_name']],
    on='id',
    how='left'
)
print ("Длина датафрейма ferma_main после добавления информации из датафрейма ferma_dad: ", len(ferma_main))

display(ferma_main.head(5))
````

Сделаем для ferma_main столбец id_father, чтобы он соответствовал уже созданным в cow_buy.
То есть — одинаковые отцы (по имени и породе) должны иметь одинаковый id_father в обеих таблицах.

````python
# 1. Создаём справочник отцов из cow_buy
father_dict = (
    cow_buy[['father_breed', 'father_name', 'id_father']]
    .drop_duplicates()
    .reset_index(drop=True)
)

# 2. Присоединяем к ferma_main по двум полям
ferma_main = ferma_main.merge(
    father_dict,
    on=['father_breed', 'father_name'],
    how='left'
)

# найти пропуски и выдать новые id, начиная с максимального
max_id = father_dict['id_father'].max()

new_fathers = (
    ferma_main[ferma_main['id_father'].isna()][['father_breed', 'father_name']]
    .drop_duplicates()
    .reset_index(drop=True)
)

new_fathers['id_father'] = range(max_id + 1, max_id + 1 + len(new_fathers))

# объединить с базовым справочником
father_dict_full = pd.concat([father_dict, new_fathers], ignore_index=True).drop_duplicates()

# обновить ferma_main окончательно
ferma_main = ferma_main.drop(columns=['id_father'], errors='ignore').merge(
    father_dict_full,
    on=['father_breed', 'father_name'],
    how='left'
)

ferma_main['id_father'] = (
    ferma_main['id_father']
    .fillna(0)          # заполняем NaN
    .astype(int)        # переводим в целые
)

father_dict_full = (
    father_dict_full
    .sort_values('id_father')      # сортируем по id_father
    .reset_index(drop=True)        # сбрасываем старый индекс и создаём новый от 0
)
````

````python
def plot_boxplots(
    df,
    columns,
    x_size=9,
    y_size=4,
    title='Boxplot количественных признаков'
):
    # выбираем только существующие и непустые колонки
    num_cols = [c for c in columns if c in df.columns and df[c].notna().any()]

    # переводим данные в long-format
    data_long = df[num_cols].melt(
        var_name='Признак',
        value_name='Значение'
    )

    plt.figure(figsize=(x_size, y_size))

    sns.boxplot(
        data=data_long,
        x='Значение',
        y='Признак',
        color='lightyellow',
        linewidth=1.5,
        fliersize=4
    )

    plt.title(title)
    plt.xlabel('Значение')
    plt.ylabel('Признак')
    plt.grid(axis='x', alpha=0.3)
    plt.tight_layout()
    plt.show()
````

````python
def plot_dist_and_box(df, columns, figsize=(12, 4)):
    for col in columns:
        fig, axes = plt.subplots(1, 2, figsize=figsize)

        # --- Гистограмма + KDE ---
        sns.histplot(
            df[col],
            kde=True,
            ax=axes[0],
            color='steelblue'
        )
        axes[0].axvline(df[col].mean(), color='red', linestyle='--', label='Среднее')
        axes[0].set_title(f'Распределение {col}')
        axes[0].set_xlabel(col)
        axes[0].set_ylabel('Частота')
        axes[0].legend()
        axes[0].grid(alpha=0.3)

        # --- Boxplot ---
        sns.boxplot(
            x=df[col],
            ax=axes[1],
            color='lightgray'
        )
        axes[1].set_title(f'Boxplot {col}')
        axes[1].set_xlabel(col)
        axes[1].grid(alpha=0.3)

        plt.tight_layout()
        plt.show()
````

````python
def plot_scatter_pairgrid(df, factor, targ, cat):
    for category in cat:
        g = sns.PairGrid(
            df,
            x_vars=factor,
            y_vars=[targ],
            hue=category,
            height=4,
            aspect=1.2
        )
        
        g.map(sns.scatterplot, alpha=0.7)
        g.add_legend(title=category)
        g.fig.suptitle(f'{targ} vs признаки (разбиение по {category})', y=1.05)
        plt.show()
````

## 5. Обучение модели линейной регрессии

### 5.1 🎯 Задачи этапа (Модель №1)

- С помощью лучшей модели спрогнозируйте удой коров, которых фермер хочет купить. Данные о них находятся в датасете cow_buy.csv.

### Зададим необходимые функции:
* Функция по обучению модели простой регрессии `evaluate_linear_model(X, y, features_num, features_cat, random_state=42)`
* Функция по предсказанию по обученной модели `make_prediction(model, df, scaler, encoder, features_num, features_cat)`
* Функции для визуализации остатков:
    * `plot_residuals_hist(y_test, predictions, bins=30)`,
    * `plot_residuals_scatter(y_test, predictions)`, 
    * `plot_actual_vs_predicted(y_test, predictions)`
* Функция для печати метрик по обученной модели `print_regression_metrics(mse, rmse, mae, r2)`
* Функция для сохранения метрик по каждой модели `save_metrics(df, model_name, mse, rmse, mae, r2)`

````python
#sklearn.__version__ - 0.24.1

# Функция по оценке модели: скейлинг → OHE → объединение → обучение → метрики.
def evaluate_linear_model(X, y, features_num, features_cat, random_state=42):
    # Разделение на train/test
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, random_state=random_state
    )
    
    # Масштабирование численных признаков
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train[features_num])
    X_test_scaled  = scaler.transform(X_test[features_num])
    
    # OHE для категориальных признаков
    encoder = OneHotEncoder(drop=None, sparse=False, handle_unknown='ignore')
    X_train_ohe = encoder.fit_transform(X_train[features_cat])
    X_test_ohe  = encoder.transform(X_test[features_cat])

    # Имена колонок
    try:
        encoder_col_names = encoder.get_feature_names_out(features_cat)
    except:
        encoder_col_names = encoder.get_feature_names(features_cat)
    
    # Превращаем в DataFrame
    X_train_scaled_df = pd.DataFrame(
        X_train_scaled, columns=features_num, index=X_train.index
    )
    X_test_scaled_df = pd.DataFrame(
        X_test_scaled, columns=features_num, index=X_test.index
    )
    
    X_train_ohe_df = pd.DataFrame(
        X_train_ohe, columns=encoder_col_names, index=X_train.index
    )
    X_test_ohe_df = pd.DataFrame(
        X_test_ohe, columns=encoder_col_names, index=X_test.index
    )
    
    # Объединение признаков
    X_train_final = pd.concat([X_train_scaled_df, X_train_ohe_df], axis=1)
    X_test_final  = pd.concat([X_test_scaled_df,  X_test_ohe_df], axis=1)
    
    # Обучение модели
    model = LinearRegression()
    model.fit(X_train_final, y_train)
    
    # Предсказания
    predictions = model.predict(X_test_final)
    
    # --- Метрики ---
    mse = round(mean_squared_error(y_test, predictions))
    rmse = round(np.sqrt(mse))
    mae = round(mean_absolute_error(y_test, predictions))
    r2 = round(r2_score(y_test, predictions), 3)

    # --- Доверительный интервал ---
    # Остатки на тренировке
    train_pred = model.predict(X_train_final)
    residuals = y_train - train_pred

    # 95% квантили
    ci_low = round(residuals.quantile(0.025))
    ci_high = round(residuals.quantile(0.975))

    # --- Коэффициенты модели ---
    coef_df = pd.DataFrame({
        'feature': X_train_final.columns,
        'coef': model.coef_,
        'abs_coef': np.abs(model.coef_)
    }).sort_values('abs_coef', ascending=False)
    
    return model, scaler, encoder, mse, rmse, mae, r2, ci_low, ci_high, coef_df, y_test, predictions, X_test_final

#Функция по предсказанию датафрейма по модели
def make_prediction(model, df, scaler, encoder, features_num, features_cat):
    # 1. Масштабирование численных признаков
    scaled = scaler.transform(df[features_num])
    scaled_df = pd.DataFrame(
        scaled,
        columns=features_num,
        index=df.index
    )

    # 2. OHE для категориальных признаков
    ohe = encoder.transform(df[features_cat])

    try:
        encoder_col_names = encoder.get_feature_names_out(features_cat)
    except:
        encoder_col_names = encoder.get_feature_names(features_cat)

    ohe_df = pd.DataFrame(
        ohe,
        columns=encoder_col_names,
        index=df.index
    )

    # 3. Собираем финальный набор признаков — ТОЧНО как в evaluate_linear_model
    X_final = pd.concat([scaled_df, ohe_df], axis=1)

    # 4. Предсказание
    df["predicted_milk"] = model.predict(X_final)

    return df, X_final

# Функцию построения графика остатков - гистограмма.    
def plot_residuals_hist(y_test, predictions, bins=30):
    residuals = y_test - predictions
    
    plt.figure(figsize=(8, 5))
    plt.hist(residuals, bins=bins, edgecolor='black')
    plt.title("Распределение остатков")
    plt.xlabel("Остаток")
    plt.ylabel("Количество")
    plt.show()

# Функцию построения графика остатков - диаграмма рассеивания.    
def plot_residuals_scatter(y_test, predictions):
    # Остатки
    residuals = y_test - predictions
    
    # График
    plt.figure(figsize=(8, 5))
    plt.scatter(predictions, residuals, alpha=0.5)
    plt.axhline(0, color='red', linestyle='--')
    plt.title("График остатков")
    plt.xlabel("Предсказанное значение")
    plt.ylabel("Остаток")
    plt.show()

# Функция plot_actual_vs_predicted
def plot_actual_vs_predicted(y_test, predictions):
    plt.figure(figsize=(8, 5))
    
    # Scatter
    plt.scatter(y_test, predictions, alpha=0.6)
    
    # Диагональная линия идеальных предсказаний
    min_val = min(y_test.min(), predictions.min())
    max_val = max(y_test.max(), predictions.max())
    plt.plot([min_val, max_val],
             [min_val, max_val],
             color='red', linestyle='--')
    
    # Подписи
    plt.xlabel("Фактическое значение")
    plt.ylabel("Предсказанное значение")
    plt.title("Фактические vs предсказанные значения")
    
    plt.show()
    
# Функция печати метрик
def print_regression_metrics(mse, rmse, mae, r2, ci_low=None, ci_high=None):
    print(f"\033[1mНа тестовой выборке:\033[0m")
    print(f"MSE  : {mse:.0f}")
    print(f"RMSE : {rmse:.0f}")
    print(f"MAE  : {mae:.0f}")
    print(f"R²   : {r2:.3f}")
    
    if (ci_low is not None) and (ci_high is not None):
        print(f"95% интервал ошибки: [{ci_low:.0f}; {ci_high:.0f}]")

# Функция по сохранению метрик в отдельный датафрейм
def save_metrics(df, model_name, mse, rmse, mae, r2, ci_low=None, ci_high=None):
    new_row = pd.DataFrame({
        'model': [model_name],
        'MSE': [mse],
        'RMSE': [rmse],
        'MAE': [mae],
        'R2': [r2],
        'CI_low': [ci_low],
        'CI_high': [ci_high]
    })
    
    df = pd.concat([df, new_row], ignore_index=True)
    return df

# Функция для построения residplot
def plot_residuals(y_true, y_pred, title='Residual plot (остатки модели)'):
    plt.figure(figsize=(8, 5))
    
    sns.residplot(
        x=y_true,
        y=y_pred,
        lowess=True,
        line_kws={'color': 'red'}
    )
    
    plt.axhline(0, color='black', linestyle='--', alpha=0.7)
    plt.title(title)
    plt.xlabel('Фактическое значение')
    plt.ylabel('Остатки (y_true − y_pred)')
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

# Функция для визуализации важности признаков по коэффициентам линейной модели
def plot_top_coefficients(
    coef_df,
    top_n=10,
    title='Топ признаков по важности (коэффициенты LinearRegression)',
    figsize=(8, 5)
):
    top = (
        coef_df
        .head(top_n)
        .sort_values('abs_coef')
    )
    
    plt.figure(figsize=figsize)
    plt.barh(top['feature'], top['coef'])
    plt.axvline(0, color='black', linewidth=0.8)
    plt.title(title)
    plt.xlabel('Коэффициент (после StandardScaler)')
    plt.ylabel('Признак')
    plt.tight_layout()
    plt.show()
````

````python
# Функция по оценке модели: скейлинг → OHE → объединение → обучение → метрики.
def evaluate_logistic_model(X, y, features_num, features_cat, random_state=42):
    # Разделение на train/test
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, random_state=random_state
    )
    
    # Масштабирование численных признаков
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train[features_num])
    X_test_scaled  = scaler.transform(X_test[features_num])
    
    # OHE для категориальных признаков
    encoder = OneHotEncoder(drop=None, sparse=False, handle_unknown='ignore')
    X_train_ohe = encoder.fit_transform(X_train[features_cat])
    X_test_ohe  = encoder.transform(X_test[features_cat])

    # Имена колонок
    try:
        encoder_col_names = encoder.get_feature_names_out(features_cat)
    except:
        encoder_col_names = encoder.get_feature_names(features_cat)
    
    # Превращаем в DataFrame
    X_train_scaled_df = pd.DataFrame(
        X_train_scaled, columns=features_num, index=X_train.index
    )
    X_test_scaled_df = pd.DataFrame(
        X_test_scaled, columns=features_num, index=X_test.index
    )
    
    X_train_ohe_df = pd.DataFrame(
        X_train_ohe, columns=encoder_col_names, index=X_train.index
    )
    X_test_ohe_df = pd.DataFrame(
        X_test_ohe, columns=encoder_col_names, index=X_test.index
    )
    
    # Объединение признаков
    X_train_final = pd.concat([X_train_scaled_df, X_train_ohe_df], axis=1)
    X_test_final  = pd.concat([X_test_scaled_df,  X_test_ohe_df], axis=1)
    
    # Обучение модели
    # создаём объект для работы с моделью
    model = LogisticRegression()
    model.fit(X_train_final, y_train)
    
    # Предсказания
    predictions = model.predict(X_test_final)
    
    # --- Метрики ---
    acc = accuracy_score(y_test, predictions)
    recall = recall_score(y_test, predictions, pos_label='вкусно')
    precision = precision_score(y_test, predictions, pos_label='вкусно')

        # --- Коэффициенты модели ---
def evaluate_logistic_model(X, y, features_num, features_cat, random_state=42):
    # Разделение на train/test
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, random_state=random_state
    )
    
    # Масштабирование численных признаков
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train[features_num])
    X_test_scaled  = scaler.transform(X_test[features_num])
    
    # OHE для категориальных признаков
    encoder = OneHotEncoder(drop=None, sparse=False, handle_unknown='ignore')
    X_train_ohe = encoder.fit_transform(X_train[features_cat])
    X_test_ohe  = encoder.transform(X_test[features_cat])

    # Имена колонок
    try:
        encoder_col_names = encoder.get_feature_names_out(features_cat)
    except:
        encoder_col_names = encoder.get_feature_names(features_cat)
    
    # Превращаем в DataFrame
    X_train_scaled_df = pd.DataFrame(
        X_train_scaled, columns=features_num, index=X_train.index
    )
    X_test_scaled_df = pd.DataFrame(
        X_test_scaled, columns=features_num, index=X_test.index
    )
    
    X_train_ohe_df = pd.DataFrame(
        X_train_ohe, columns=encoder_col_names, index=X_train.index
    )
    X_test_ohe_df = pd.DataFrame(
        X_test_ohe, columns=encoder_col_names, index=X_test.index
    )
    
    # Объединение признаков
    X_train_final = pd.concat([X_train_scaled_df, X_train_ohe_df], axis=1)
    X_test_final  = pd.concat([X_test_scaled_df,  X_test_ohe_df], axis=1)
    
    # Обучение модели
    # создаём объект для работы с моделью
    model = LogisticRegression()
    model.fit(X_train_final, y_train)
    
    # Предсказания
    predictions = model.predict(X_test_final)
    
    # --- Метрики ---
    acc = accuracy_score(y_test, predictions)
    recall = recall_score(y_test, predictions, pos_label='вкусно')
    precision = precision_score(y_test, predictions, pos_label='вкусно')

        # --- Коэффициенты модели ---
    coef_df = pd.DataFrame({
        'feature': X_train_final.columns,
        'coef': model.coef_[0],              
        'abs_coef': np.abs(model.coef_[0])
    }).sort_values('abs_coef', ascending=False)
    
    return model, scaler, encoder, acc, recall, precision, coef_df, y_test, predictions, X_test_final, X_train_final   

def make_prediction_logistic(model, df, scaler, encoder, features_num, features_cat,
                             pos_label='вкусно', threshold=0.5):
    # 1) Масштабирование численных признаков
    scaled = scaler.transform(df[features_num])
    scaled_df = pd.DataFrame(scaled, columns=features_num, index=df.index)

    # 2) OHE для категориальных признаков
    ohe = encoder.transform(df[features_cat])

    try:
        encoder_col_names = encoder.get_feature_names_out(features_cat)
    except:
        encoder_col_names = encoder.get_feature_names(features_cat)

    ohe_df = pd.DataFrame(ohe, columns=encoder_col_names, index=df.index)

    # 3) Финальный набор признаков (как при обучении)
    X_final = pd.concat([scaled_df, ohe_df], axis=1)

    # 4) Вероятности: берём именно pos_label (важно!)
    pos_index = list(model.classes_).index(pos_label)
    proba_pos = model.predict_proba(X_final)[:, pos_index]

    # 5) Классификация по порогу
    pred_label = np.where(proba_pos >= threshold, pos_label, f'не_{pos_label}')

    # 6) Записываем в df
    df = df.copy()
    df["proba_vkusno"] = proba_pos
    df["predicted_taste"] = pred_label

    return df, X_final### Зададим необходимые функции:

# Функция печати метрик
def print_logistic_metrics(acc, recall, precision):
    print(f"\033[1mНа тестовой выборке:\033[0m")
    print(f"Accuracy score  : {acc:.3f}")
    print(f"Recall : {recall:.3f}")
    print(f"Precision  : {precision:.3f}")

def print_coef(model, X_train_final):    
    for feature_number, feature_name in enumerate(X_train_final.columns):
        weight = model.coef_[0, feature_number]
        print(f"{feature_name:<20}: {weight:+.6f}") 
````

````python
def threshold_analysis_precision_full(y_test, y_proba, pos_label='вкусно'):
    
    y_true_bin = (y_test == pos_label).astype(int)
    precisions, recalls, thresholds = precision_recall_curve(y_true_bin, y_proba)

    rows = []

    for i, thr in enumerate(thresholds):
        y_pred_thr = (y_proba >= thr).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_true_bin,
            y_pred_thr
        ).ravel()

        rows.append({
            "Порог": round(thr, 3),
            "precision": round(precisions[i], 3),
            "recall": round(recalls[i], 3),
            "Количество «правильных единичек» (TP)": tp,
            "Количество False Positive (FP)": fp,
            "Количество False Negative (FN)": fn
        })

    return pd.DataFrame(rows)


def plot_precision_recall_vs_threshold(y_true, y_proba, thresholds=None):
    if thresholds is None:
        thresholds = np.linspace(0, 1, 200)

    precisions = []
    recalls = []

    for t in thresholds:
        y_pred = (y_proba >= t).astype(int)
        precisions.append(precision_score(y_true, y_pred, zero_division=0))
        recalls.append(recall_score(y_true, y_pred, zero_division=0))

    plt.figure(figsize=(12, 6))
    plt.plot(thresholds, precisions, label="Precision", linewidth=2)
    plt.plot(thresholds, recalls, label="Recall", linewidth=2)
    plt.xlabel("Порог классификации")
    plt.ylabel("Значение метрики")
    plt.title("Зависимость Precision и Recall от порога классификации")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()
````
